# Q1 — UNet Autoencoder tests

Run cells top to bottom. Requires the files in this folder and `adl-cpu` env.

````python
import torch
from model import UNetAE, add_gaussian_noise

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('device:', device)

model = UNetAE(in_channels=1, base_ch=32).to(device)
x = torch.rand(4, 1, 28, 28, device=device)
y = model(x)
print('output shape:', tuple(y.shape))
print('output range:', y.min().item(), y.max().item())
print('params (M):', round(sum(p.numel() for p in model.parameters()) / 1e6, 3))
assert y.shape == x.shape
print('shape OK')
````

````python
noisy = add_gaussian_noise(x, sigma=0.2)
print('noisy shape:', tuple(noisy.shape))
print('noisy range:', noisy.min().item(), noisy.max().item())
assert noisy.shape == x.shape
print('noise OK')
````
