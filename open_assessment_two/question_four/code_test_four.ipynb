```python
import matplotlib.pyplot as plt
from data import get_train_captions, build_vocab, Flickr30kDataset

captions = get_train_captions(dataset_name="AnyModal/flickr30k", max_samples=200)
stoi, itos = build_vocab(captions, min_freq=1)
print("vocab size:", len(itos))

ds = Flickr30kDataset(
    split="train",
    stoi=stoi,
    max_samples=10,
    img_size=64,
    dataset_name="AnyModal/flickr30k",
)

n = 4
fig, axes = plt.subplots(1, n, figsize=(12, 3))
for i in range(n):
    img, ids = ds[i]
    words = [itos[idx] for idx in ids.tolist() if idx != stoi["<pad>"]]
    caption = " ".join(words)
    axes[i].imshow(img.permute(1, 2, 0))
    axes[i].set_title(caption, fontsize=8)
    axes[i].axis("off")
plt.tight_layout()
plt.show()
```
